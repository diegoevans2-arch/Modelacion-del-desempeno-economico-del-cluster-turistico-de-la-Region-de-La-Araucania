```python
import pyreadr
import pandas as pd

# Ruta del archivo
ruta = r"C:\Users\diego.perezp\Documents\BBDD_tesis_doctorado_german\inversiones_202503.rdata"

# Cargar el archivo .rdata
# read_r() retorna un diccionario con todos los objetos R
objetos = pyreadr.read_r(ruta)

# Ver qué objetos contiene el archivo
print("Objetos en el .rdata:")
print(objetos.keys())
print("\n" + "="*60 + "\n")

# Explorar el contenido de cada objeto
for nombre, objeto in objetos.items():
    print(f"Objeto: {nombre}")
    print(f"Tipo: {type(objeto)}")
    if isinstance(objeto, pd.DataFrame):
        print(f"Shape: {objeto.shape}")
        print(f"Primeras filas:\n{objeto.head()}\n")
    else:
        print(f"Contenido: {objeto}\n")
    print("-"*60 + "\n")
```

```python
# Extrae el dataframe (asume que el primer objeto es el que necesitas)
# Si tienes varios objetos, ajusta el nombre según lo que viste en Celda 2

nombre_dataframe = list(objetos.keys())[0]  # Toma el primer objeto
df = objetos[nombre_dataframe]

# Ruta de salida
ruta_salida = r"C:\Users\diego.perezp\Documents\BBDD_tesis_doctorado_german\inversiones_202503.csv"

# Guardar como CSV
df.to_csv(ruta_salida, index=False, encoding='utf-8')

print(f"✓ Archivo convertido exitosamente")
print(f"Guardado en: {ruta_salida}")
print(f"Filas: {df.shape[0]} | Columnas: {df.shape[1]}")
```

```python
# Verificar que el CSV se creó correctamente
df_validacion = pd.read_csv(ruta_salida)
print("Validación del CSV:")
print(df_validacion.head())
print(f"\nShape: {df_validacion.shape}")
```

```python
# Columnas a mostrar
columnas = [
    'EBI_ANO_POSTULA',
    'EBI_ANO_CONVENIO',
    'EBI_GASTO_ANTERIOR',
    'EBI_ULT_ANO_INV',
    'EBI_GASTO_VIGENTE',
    'EBI_VIDA_UTIL',
    'REG_CLAVE'
]

# Validación con solo las columnas especificadas
print("Validación del CSV (columnas seleccionadas):\n")
print(df_validacion[columnas].head())
print(f"\nShape total: {df_validacion[columnas].shape}")
```
