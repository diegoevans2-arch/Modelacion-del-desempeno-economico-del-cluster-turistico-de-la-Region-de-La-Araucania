## 1️⃣ IMPORTS Y CONFIGURACIÓN

```python
import pandas as pd
import numpy as np
from pathlib import Path

# Configuración de pandas para visualización
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
pd.set_option('display.width', None)
pd.set_option('display.float_format', '{:.2f}'.format)

print("✅ Librerías cargadas correctamente")
print(f"   - Pandas version: {pd.__version__}")
print(f"   - NumPy version: {np.__version__}")
```

---
## 2️⃣ LECTURA Y DIAGNÓSTICO INICIAL

```python
# Ruta del archivo en Windows
archivo_emat = r"C:\Users\diego.perezp\Documents\BBDD_tesis_doctorado_german\consolidado_ReporteDinamicoEMAT_2017_2026.xlsx"

# Verificar que el archivo existe
if not Path(archivo_emat).exists():
    print(f"❌ ERROR: El archivo no existe en la ruta especificada:")
    print(f"   {archivo_emat}")
    print(f"\n💡 Verifica la ruta y vuelve a ejecutar esta celda")
else:
    print(f"✅ Archivo encontrado: {Path(archivo_emat).name}")
    print(f"   Tamaño: {Path(archivo_emat).stat().st_size / 1024:.2f} KB")
```

```python
# Lectura inicial sin procesar
df_raw = pd.read_excel(archivo_emat, sheet_name='Hoja1')

print("=" * 80)
print("📊 DIAGNÓSTICO INICIAL DEL ARCHIVO")
print("=" * 80)
print(f"\n1. Dimensiones originales: {df_raw.shape[0]} filas × {df_raw.shape[1]} columnas")
print(f"\n2. Primeras 3 filas (estructura cruda):")
display(df_raw.head(3))
print(f"\n3. Nombres de columnas actuales:")
print(df_raw.columns.tolist())
```

---
## 3️⃣ ANÁLISIS DE DUPLICADOS (FILAS '--' vs '1')

**Objetivo:** Verificar si las filas con `N. = '--'` y `N. = '1'` son duplicadas exactas o tienen diferencias.

```python
# Extraer solo las filas con datos (excluir fila 0 que es encabezado)
df_datos = df_raw.iloc[1:].copy()
df_datos.reset_index(drop=True, inplace=True)

# Separar filas con N. = '--' y N. = '1'
filas_guion = df_datos[df_datos.iloc[:, 1] == '--'].copy()
filas_uno = df_datos[df_datos.iloc[:, 1] == 1].copy()

print("=" * 80)
print("🔍 ANÁLISIS DE DUPLICADOS: FILAS '--' vs FILAS '1'")
print("=" * 80)
print(f"\n1. Número de filas con N. = '--': {len(filas_guion)}")
print(f"   Número de filas con N. = '1': {len(filas_uno)}")

# Comparar valores numéricos entre ambos grupos por año
print(f"\n2. Comparación de valores por año:")
print(f"   (Verificando si las filas -- y 1 son idénticas)")

# Para cada año, comparar si los valores son idénticos
anos_unicos = df_datos.iloc[:, 0].unique()
anos_unicos = [a for a in anos_unicos if a != 'AÑO']  # Excluir el encabezado

son_identicos = True
diferencias = []

for ano in anos_unicos:
    fila_guion = df_datos[(df_datos.iloc[:, 0] == ano) & (df_datos.iloc[:, 1] == '--')]
    fila_uno = df_datos[(df_datos.iloc[:, 0] == ano) & (df_datos.iloc[:, 1] == 1)]
    
    if len(fila_guion) > 0 and len(fila_uno) > 0:
        # Comparar todas las columnas numéricas (índices con valores totales)
        cols_numericas = [7, 13, 19, 25, 31, 37, 43]
        
        for col_idx in cols_numericas:
            val_guion = fila_guion.iloc[0, col_idx]
            val_uno = fila_uno.iloc[0, col_idx]
            
            # Comparar considerando NaN
            if pd.isna(val_guion) and pd.isna(val_uno):
                continue
            elif pd.isna(val_guion) or pd.isna(val_uno):
                son_identicos = False
                diferencias.append(f"Año {ano}, col {col_idx}: guion={val_guion}, uno={val_uno}")
            elif val_guion != val_uno:
                son_identicos = False
                diferencias.append(f"Año {ano}, col {col_idx}: guion={val_guion}, uno={val_uno}")

if son_identicos:
    print(f"\n   ✅ CONFIRMACIÓN: Las filas '--' y '1' son IDÉNTICAS")
    print(f"   ✅ Las filas '--' NO son totales agregados, son duplicados exactos")
    print(f"   ✅ Decisión: Mantener solo filas con N. = '1' (La Araucanía)")
else:
    print(f"\n   ⚠️  ALERTA: Se encontraron diferencias entre filas '--' y '1':")
    for dif in diferencias[:10]:  # Mostrar las primeras 10
        print(f"      {dif}")
    print(f"\n   📌 Interpretación: Las filas '--' parecen ser agregados o valores redondeados")
    print(f"   📌 Las filas '1' son valores específicos de La Araucanía")
    print(f"   ✅ Decisión: Mantener solo filas con N. = '1' (región específica)")
```

---
## 4️⃣ CONSTRUCCIÓN DE NOMBRES DE COLUMNAS

**Estrategia:** Combinar nombres principales + subencabezados de la fila 0 para crear nombres descriptivos y únicos.

```python
print("=" * 80)
print("🏗️  CONSTRUCCIÓN DE NOMBRES DE COLUMNAS")
print("=" * 80)

# La fila 0 contiene los sub-encabezados (Chi., Ext., Total, Var%)
fila_subencabezados = df_raw.iloc[0].tolist()

# Nombres principales de columnas
nombres_principales = df_raw.columns.tolist()

# Construir nombres de columnas finales combinando principal + subencabezado
nombres_finales = []
principal_actual = None  # Para rastrear el nombre principal actual

for i, (principal, subencabezado) in enumerate(zip(nombres_principales, fila_subencabezados)):
    # Las primeras 3 columnas son metadatos (AÑO, N., AGRUPACIÓN)
    if i < 3:
        if i == 0:
            nombres_finales.append('AÑO')
        elif i == 1:
            nombres_finales.append('FRECUENCIA')
        elif i == 2:
            nombres_finales.append('AGRUPACION')
    else:
        # Actualizar el principal si no es Unnamed
        if 'Unnamed' not in str(principal):
            principal_actual = str(principal).strip().upper()
            principal_actual = principal_actual.replace(' ', '_')
            principal_actual = principal_actual.replace('Ó', 'O')
            principal_actual = principal_actual.replace('Í', 'I')
        
        # Limpiar el subencabezado
        sub_limpio = str(subencabezado).strip()
        
        # Mapeo de subencabezados a sufijos claros
        mapeo_sufijos = {
            'Chi.': 'Chileno',
            'Ext.': 'Extranjero',
            'Total': 'Total',
            'Var. 12 meses (%)': 'Var_12m_pct',
            'Var. 12 meses (pp)': 'Var_12m_pp',
            'Hotel': 'Hotel',
            'Otros': 'Otros'
        }
        
        sufijo = mapeo_sufijos.get(sub_limpio, sub_limpio)
        
        # Construir nombre final con estructura específica
        if sufijo == 'Var_12m_pct' or sufijo == 'Var_12m_pp':
            # La variación corresponde al valor anterior
            if len(nombres_finales) > 0:
                nombre_anterior = nombres_finales[-1]
                nombre_final = f"{nombre_anterior}_Var"
            else:
                nombre_final = f"{principal_actual}_{sufijo}"
        else:
            # Es un valor directo
            nombre_final = f"{principal_actual}_{sufijo}"
        
        nombres_finales.append(nombre_final)

print(f"\n1. Nombres de columnas generados: {len(nombres_finales)}")
print(f"\n2. Primeras 10 columnas finales:")
for i, nombre in enumerate(nombres_finales[:10]):
    print(f"   {i}: {nombre}")

print(f"\n3. Últimas 10 columnas finales:")
for i, nombre in enumerate(nombres_finales[-10:], start=len(nombres_finales)-10):
    print(f"   {i}: {nombre}")
```

---
## 5️⃣ LIMPIEZA Y TRANSFORMACIÓN

**Acción:** Mantener solo filas con `N. = 1` (La Araucanía) y aplicar los nuevos nombres de columnas.

```python
print("=" * 80)
print("🧹 LIMPIEZA Y TRANSFORMACIÓN DE DATOS")
print("=" * 80)

# Tomar solo las filas con N. = 1 (ya confirmamos que -- tienen diferencias)
df_clean = df_datos[df_datos.iloc[:, 1] == 1].copy()

# Asignar los nuevos nombres de columnas
df_clean.columns = nombres_finales

# Resetear índice
df_clean.reset_index(drop=True, inplace=True)

print(f"\n1. Dimensiones después de limpieza: {df_clean.shape[0]} filas × {df_clean.shape[1]} columnas")
print(f"\n2. Primeras 5 filas del DataFrame limpio:")
display(df_clean.head())
```

---
## 6️⃣ CONVERSIÓN DE TIPOS DE DATOS

**Objetivo:** Convertir columnas a sus tipos correctos (int, float, string) para facilitar análisis posterior.

```python
print("=" * 80)
print("🔄 CONVERSIÓN DE TIPOS DE DATOS")
print("=" * 80)

# Convertir AÑO a entero
df_clean['AÑO'] = pd.to_numeric(df_clean['AÑO'], errors='coerce').astype('Int64')

# Convertir FRECUENCIA a string
df_clean['FRECUENCIA'] = df_clean['FRECUENCIA'].astype(str)

# Convertir AGRUPACION a string
df_clean['AGRUPACION'] = df_clean['AGRUPACION'].astype(str)

# Verificar si hay columnas duplicadas y hacerlas únicas si es necesario
columnas_duplicadas = df_clean.columns[df_clean.columns.duplicated()].unique()
if len(columnas_duplicadas) > 0:
    print(f"\n⚠️  ALERTA: Se encontraron {len(columnas_duplicadas)} columnas con nombres duplicados:")
    for col_dup in columnas_duplicadas:
        print(f"   - {col_dup}")
    print("\n   Añadiendo sufijos numéricos para hacerlas únicas...")
    
    # Hacer los nombres únicos añadiendo sufijos
    columnas_nuevas = []
    contador_por_nombre = {}
    
    for col in df_clean.columns:
        if col in contador_por_nombre:
            contador_por_nombre[col] += 1
            columnas_nuevas.append(f"{col}_{contador_por_nombre[col]}")
        else:
            contador_por_nombre[col] = 0
            columnas_nuevas.append(col)
    
    df_clean.columns = columnas_nuevas
    print(f"   ✅ Nombres de columnas ahora son únicos")

# Convertir todas las columnas numéricas (del índice 3 en adelante) a float
print(f"\n1. Convirtiendo columnas numéricas...")
for col in df_clean.columns[3:]:
    try:
        df_clean[col] = pd.to_numeric(df_clean[col], errors='coerce')
    except Exception as e:
        print(f"   ⚠️  Error al convertir columna {col}: {e}")

print(f"\n2. Tipos de datos después de conversión:")
print(df_clean.dtypes)

print(f"\n3. Verificación de valores nulos por columna:")
nulos_por_columna = df_clean.isnull().sum()
cols_con_nulos = nulos_por_columna[nulos_por_columna > 0]
print(f"\n   Columnas con valores nulos: {len(cols_con_nulos)}")
print(cols_con_nulos)
```

---
## 7️⃣ AGREGAR COLUMNA INDICADORA: `tiene_llegadas`

**Objetivo:** Crear una columna booleana que indique si el año tiene datos de llegadas disponibles.

```python
print("=" * 80)
print("➕ AGREGAR COLUMNA INDICADORA: tiene_llegadas")
print("=" * 80)

# Identificar columnas de llegadas (sin incluir variaciones)
cols_llegadas = [col for col in df_clean.columns if 'LLEGADAS' in col and 'Var' not in col]

print(f"\n1. Columnas de llegadas identificadas:")
for col in cols_llegadas:
    print(f"   - {col}")

# Crear columna indicadora: True si tiene al menos un valor no nulo en llegadas
df_clean['tiene_llegadas'] = df_clean[cols_llegadas].notna().any(axis=1)

print(f"\n2. Distribución de registros con/sin llegadas:")
print(df_clean['tiene_llegadas'].value_counts())

print(f"\n3. Años con datos de llegadas:")
print(df_clean[df_clean['tiene_llegadas']]['AÑO'].tolist())

print(f"\n4. Años SIN datos de llegadas:")
print(df_clean[~df_clean['tiene_llegadas']]['AÑO'].tolist())
```

---
## 8️⃣ VALIDACIÓN Y EXPLORACIÓN FINAL

**Verificaciones:**
- Dimensiones finales
- Tipos de datos
- Estadísticos descriptivos
- Completitud de datos

```python
print("=" * 80)
print("✅ VALIDACIÓN Y EXPLORACIÓN FINAL")
print("=" * 80)

print(f"\n1. DIMENSIONES FINALES: {df_clean.shape[0]} filas × {df_clean.shape[1]} columnas")

print(f"\n2. INFORMACIÓN DEL DATAFRAME:")
df_clean.info()

print(f"\n3. ESTADÍSTICOS DESCRIPTIVOS:")
display(df_clean.describe())

print(f"\n4. MUESTRA DE DATOS (primeras 5 filas):")
display(df_clean.head())

print(f"\n5. RESUMEN DE COMPLETITUD DE DATOS:")
completitud = (1 - df_clean.isnull().sum() / len(df_clean)) * 100
print(f"\n   Columnas con 100% de datos completos:")
cols_completas = completitud[completitud == 100].index.tolist()
print(f"   Total: {len(cols_completas)} columnas")
for col in cols_completas:
    print(f"   - {col}")

print(f"\n   Columnas con datos faltantes (ordenadas por % completitud):")
cols_incompletas = completitud[completitud < 100].sort_values()
print(f"   Total: {len(cols_incompletas)} columnas")
for col, pct in cols_incompletas.head(15).items():
    print(f"   - {col}: {pct:.1f}% completo")
```

---
## 9️⃣ GUARDAR A CSV

**Output:** DataFrame limpio guardado en formato CSV para análisis posterior.

```python
print("=" * 80)
print("💾 GUARDAR DATAFRAME A CSV")
print("=" * 80)

# Definir ruta de salida en la misma carpeta del archivo original
directorio_salida = Path(archivo_emat).parent
archivo_salida = directorio_salida / 'emat_transformado_2017_2026.csv'

# Guardar a CSV
df_clean.to_csv(archivo_salida, index=False, encoding='utf-8')

print(f"\n✅ Archivo guardado exitosamente en:")
print(f"   {archivo_salida}")
print(f"\n📊 Resumen del archivo guardado:")
print(f"   - Filas: {df_clean.shape[0]}")
print(f"   - Columnas: {df_clean.shape[1]}")
print(f"   - Tamaño: {archivo_salida.stat().st_size / 1024:.2f} KB")

print("\n" + "=" * 80)
print("🎉 TRANSFORMACIÓN COMPLETADA EXITOSAMENTE")
print("=" * 80)
print(f"\n📌 DataFrame final disponible en variable: df_clean")
print(f"📌 Archivo CSV disponible en: {archivo_salida}")
```

## 📊 ANÁLISIS EXPLORATORIO ADICIONAL (OPCIONAL)

Puedes ejecutar las siguientes celdas para hacer análisis adicionales.

```python
# Evolución temporal de Llegadas Totales
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))
plt.plot(df_clean['AÑO'], df_clean['LLEGADAS_Total'], marker='o', linewidth=2, markersize=8)
plt.title('Evolución de Llegadas Totales - La Araucanía (2017-2026)', fontsize=14, fontweight='bold')
plt.xlabel('Año', fontsize=12)
plt.ylabel('Llegadas Totales', fontsize=12)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# Mostrar estadísticas
print("\n📈 Estadísticas de Llegadas Totales:")
print(f"   Máximo: {df_clean['LLEGADAS_Total'].max():,.0f} ({df_clean[df_clean['LLEGADAS_Total'] == df_clean['LLEGADAS_Total'].max()]['AÑO'].values[0]})")
print(f"   Mínimo: {df_clean['LLEGADAS_Total'].min():,.0f} ({df_clean[df_clean['LLEGADAS_Total'] == df_clean['LLEGADAS_Total'].min()]['AÑO'].values[0]})")
print(f"   Promedio: {df_clean['LLEGADAS_Total'].mean():,.0f}")
print(f"   Caída 2019-2020: {((df_clean[df_clean['AÑO']==2020]['LLEGADAS_Total'].values[0] / df_clean[df_clean['AÑO']==2019]['LLEGADAS_Total'].values[0]) - 1) * 100:.1f}%")
```

```python
# Análisis de correlaciones entre métricas principales
metricas_principales = [
    'LLEGADAS_Total',
    'PERNOCTACIONES_Total',
    'ESTANCIA_MEDIA_Total',
    'TASA_DE_OCUPACION_EN_HABITACIONES_Total',
    'TASA_DE_OCUPACION_EN_PLAZAS_Total',
    'REVPAR_Total',
    'ADR_Total'
]

# Matriz de correlación
correlaciones = df_clean[metricas_principales].corr()

print("\n🔗 MATRIZ DE CORRELACIONES ENTRE MÉTRICAS PRINCIPALES:")
print("="*80)
display(correlaciones.round(3))

# Visualizar con heatmap
plt.figure(figsize=(10, 8))
import seaborn as sns
sns.heatmap(correlaciones, annot=True, fmt='.2f', cmap='coolwarm', center=0, 
            square=True, linewidths=1, cbar_kws={"shrink": 0.8})
plt.title('Correlaciones entre Métricas EMAT', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```
